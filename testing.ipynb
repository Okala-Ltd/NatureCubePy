```python
import pandas as pd
def get_camera_trap_data(
    hdr: dict[str, str],
    datatype: str = "image",
):
    stations = get_station_info(hdr, datatype)

    if "project_system_record_id" not in stations.columns:
        raise ValueError(
            "Station data does not contain 'project_system_record_id' column. "
            "Cannot link media assets to stations."
        )

    station_ids = stations["project_system_record_id"]

    media = get_media_assets(hdr, datatype, project_system_record_ids=station_ids)
    media_segments = get_media_segments(hdr, datatype, project_system_record_ids=station_ids)

    location_df = pd.DataFrame({
        "project_system_record_id": stations["project_system_record_id"].astype(int),
        "latitude": stations.geometry.centroid.y,
        "longitude": stations.geometry.centroid.x,
    })
    df = media.merge(media_segments, on="segment_record_id", how="left")

    return df.merge(location_df, on="project_system_record_id", how="left")

```

```python
# Camera trap location pull (image + video)
import pandas as pd

camera_frames = []
for dtype in ["image", "video"]:
    gdf = get_station_info(hdr, datatype=dtype).copy()
    gdf["camera_datatype"] = dtype
    camera_frames.append(gdf)

camera_trap_locations = pd.concat(camera_frames)
```

```python
import pandas as pd

# 1) Camera lookup table (one row per camera/PSR)
camera_lookup = (
    camera_trap_locations[
        ["project_system_record_id", "device_id", "camera_datatype"]
    ]
    .dropna(subset=["project_system_record_id"])
    .drop_duplicates()
)

# Keep only image stations
image_psr = (
    camera_lookup.loc[camera_lookup["camera_datatype"] == "image", "project_system_record_id"]
    .astype(int)
    .unique()
)

# 2) Pull all image media for each PSR, and stamp PSR onto each returned row
image_frames = []
for psr_id in image_psr:
    assets = get_media_assets(hdr, "image", int(psr_id)).copy()
    assets["project_system_record_id"] = int(psr_id)
    image_frames.append(assets)

images = pd.concat(image_frames, ignore_index=True) if image_frames else pd.DataFrame()

# 3) Match each image to camera/device
images_with_camera = images.merge(
    camera_lookup[["project_system_record_id", "device_id"]],
    on="project_system_record_id",
    how="left",
)

print(f"Images: {len(images_with_camera)}")
print(f"Matched to camera: {images_with_camera['device_id'].notna().sum()}")
display(images_with_camera.head(50))
```

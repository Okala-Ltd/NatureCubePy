# NatureCubePy Authentication Tutorial

This tutorial covers how to authenticate with the Okala API using NatureCubePy.

## Prerequisites

- A valid Okala API key (obtain from your Okala dashboard)
- NatureCubePy installed (`pip install naturecubepy` or `uv add naturecubepy`)

## Import Libraries

First, let's import the necessary modules from NatureCubePy.

````python
import os
from naturecubepy import get_key, auth_headers, get_project
````

---

## Method 1: Environment Variable (Recommended)

The most secure approach is to set your API key as an environment variable.

### Setting the environment variable

**In your terminal (before starting Jupyter):**

```bash
# macOS/Linux
export OKALA_API_KEY="your_api_key_here"

# Windows (Command Prompt)
set OKALA_API_KEY=your_api_key_here

# Windows (PowerShell)
$env:OKALA_API_KEY = "your_api_key_here"
```

**Or set it in Python (for demo purposes):**

````python
# Set your API key here (replace with your actual key)
os.environ["OKALA_API_KEY"] = "your_api_key_here"  # Replace with your actual API key
````

### Using get_key() and auth_headers()

Now retrieve the key and create authentication headers:

````python
# Retrieve the API key from environment
api_key = get_key(key_name='EV2_PROD')

# Create authentication headers
hdr = auth_headers(api_key)
````

### Verify the connection

````python
# Verify authentication by getting project info
get_project(hdr)
````

---

## Method 2: Direct API Key (Quick Testing)

For quick testing, you can pass the API key directly to `auth_headers()`:

> ⚠️ **Warning:** Avoid hardcoding API keys in notebooks you share or commit to version control.

````python
# Pass the API key directly (not recommended for production)
hdr = auth_headers("your_api_key_here")

# Verify it works
get_project(hdr)
````

---

## Method 3: Using a .env File

For local development, use a `.env` file with `python-dotenv`.

### Step 1: Create a `.env` file in your project directory

```
OKALA_API_KEY=your_api_key_here
```

### Step 2: Load the environment variables

````python
api_key = get_key(key_name='OKALA_API_KEY')
hdr = auth_headers(api_key)

get_project(hdr)
````

> 💡 **Tip:** Add `.env` to your `.gitignore` to keep credentials out of version control.

## Development vs Production API

NatureCubePy supports both production and development endpoints.

````python
# Production API (default)
hdr_prod = auth_headers("your_api_key", okala_url='https://naturecube.io/api/')
print(f"Production URL: {hdr_prod['root']}")

# Development API
local_api_url = "http://localhost:8000/api/"  # Replace with your local API URL if different
hdr_dev = auth_headers("your_api_key", okala_url=local_api_url)
print(f"Development URL: {hdr_dev['root']}")
````

## Complete Example

Here's a full workflow from authentication to fetching data:

````python
import os
from naturecubepy import get_key, auth_headers, get_project, get_station_info

# Set your API key (replace with your actual key)
os.environ["OKALA_API_KEY"] = "your_api_key_here"

# Get key and authenticate
api_key = get_key("OKALA_API_KEY")
hdr = auth_headers(api_key)

# Verify authentication
get_project(hdr)
````

````python
# Now fetch station data
stations = get_station_info(hdr, datatype="video")
print(f"Found {len(stations)} stations")
stations.head()
````

## Troubleshooting

### Common Errors

| Error | Cause | Solution |
|-------|-------|----------|
| `EnvironmentError: OKALA_API_KEY environment variable not set` | `get_key()` couldn't find your API key | Set the environment variable or use `auth_headers("key")` directly |
| `httpx.HTTPStatusError: 401 Unauthorized` | Invalid or expired API key | Check the key is correct and still active |
| `httpx.HTTPStatusError: 403 Forbidden` | Key lacks access to resource | Contact your Okala administrator |

### Verify your environment variable is set

````python
# Check if the environment variable is set
api_key = os.environ.get("OKALA_API_KEY", None)

if api_key:
    print(f"✅ OKALA_API_KEY is set (starts with: {api_key[:8]}...)")
else:
    print("❌ OKALA_API_KEY is not set")
````
